```python
# Import the libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    classification_report
)
```

```python
#Creating the dataset

dataset = {
    "ingrediant" : ["apple", "bacon", "banana", "carrot", "celery", "cheese", "grapes", "greenbeans", "nuts", "orange"],
    "sweetness" : [10, 1, 10, 7, 3, 1, 8, 3, 3, 7],
    "crunchiness" : [9, 4, 1, 10, 10, 1, 5, 7, 6, 3],
    "food type" : ["fruit", "protien", "fruit", "vegie", "vegie", "protien", "fruit", "vegie", "protien", "fruit"]
}
```

```python
# Create the alias of Pandas

df = pd.DataFrame(dataset)
df.head(10)
```

```python
# df.describe() generates summary statistics for numerical columns in a DataFrame, such as count, mean, standard deviation, minimum, and maximum.
# It helps quickly understand the data's distribution and central tendency.

df.describe()
```

```python
# df.shape displays the number of rows and columns in the dataset.
# df.columns lists all column names, while df.info() shows column data types, non-null counts, and memory usage.

print("Dataset Shape:", df.shape)

print("\nColumns:\n")
print(df.columns)

print("\nDataset Information:\n")
print(df.info())
```

```python
# df.isnull().sum() counts the number of missing (NaN or null) values in each column of the DataFrame.
# It helps identify columns with missing data that may require cleaning or preprocessing.

print(df.isnull().sum())
```

```python
# selects the input features & the target variable

X = df[['sweetness', 'crunchiness']]
y = df['food type']
```

```python
# defines the tomato's features, where sweetness is 6 and crunchiness is 4.

tomato = [[6, 4]]
```

```python
# KNN classification model that uses the 4 nearest neighbors to make predictions.

model = KNeighborsClassifier(
    n_neighbors=4
    )
```

```python
# trains the KNN model using the input features (X) and their corresponding target labels (y).

model.fit(X, y)
```

```python
# predicts the food type of the tomato

y_pred = model.predict(tomato)

print(y_pred)
```
